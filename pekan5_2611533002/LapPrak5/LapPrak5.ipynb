## 2.1 Dasar Teori

### 1. Python ##

Python adalah salah satu bahasa pemrograman yang dikenal dengan sintaksnya yang sederhana sehingga cukup mudah dipahami dan digunakan untuk berbagai kebutuhan pemrograman. Pada praktikum ini, Python dimanfaatkan untuk mempraktikkan beberapa konsep dasar, seperti perulangan, percabangan, penggunaan operator, pengambilan input dari pengguna, serta pembuatan berbagai pola melalui keluaran program.



### 2. Variable

Variabel berfungsi sebagai tempat untuk menyimpan data atau nilai yang diperlukan selama program dijalankan. Dalam praktikum ini, terdapat beberapa variabel yang digunakan, seperti `batas_3002`, `ulang_3002`, `i_3002`, `j_3002`, `tinggi_3002`, dan `n_3002`.

Setiap variabel memiliki peran yang berbeda, misalnya untuk menentukan batas perulangan, menyimpan nilai penghitung, melakukan proses perhitungan, serta mengatur posisi karakter ketika membentuk suatu pola. Dengan penggunaan variabel, proses dalam program dapat dilakukan secara lebih teratur dan mudah dikendalikan.




### 3. Input dan conversi Tipe data

Fungsi `input()` digunakan untuk mengambil data yang dimasukkan oleh pengguna saat program dijalankan. Data yang diperoleh melalui `input()` pada dasarnya memiliki tipe string atau teks. Agar data tersebut dapat digunakan dalam perhitungan maupun menentukan jumlah perulangan, diperlukan fungsi `int()` untuk mengubahnya menjadi bilangan bulat.

Dalam praktikum ini, penggunaan `input()` dan `int()` diterapkan untuk menerima nilai dari pengguna, seperti batas perulangan, ukuran, atau tinggi pola yang akan dibuat oleh program.



### 4. Perulangan for

Perulangan `for` digunakan untuk menjalankan instruksi secara berulang berdasarkan urutan atau jumlah tertentu. Dalam praktikum ini, `for` dimanfaatkan untuk berbagai keperluan, seperti menampilkan angka dan karakter, melakukan proses perhitungan, serta membentuk pola tertentu.

Untuk mengatur proses perulangan, `for` dapat dipadukan dengan fungsi `range()`. Fungsi tersebut digunakan untuk menentukan nilai awal, batas perulangan, serta perubahan nilai pada setiap proses perulangan.


### 5. Fungsi range()

Fungsi `range()` digunakan untuk menghasilkan urutan bilangan yang kemudian dapat dimanfaatkan dalam perulangan `for`. Fungsi ini membantu menentukan dari nilai mana perulangan dimulai, sampai batas mana perulangan dilakukan, serta perubahan nilai pada setiap langkahnya.

Beberapa bentuk penggunaan `range()` antara lain:

```python
range(batas)
```

```python
range(awal, batas)
```

```python
range(awal, batas, langkah)
```

Dalam praktikum, `range()` digunakan dengan nilai awal 0 maupun 1 sesuai dengan kebutuhan program. Selain itu, langkah negatif juga digunakan untuk membuat perulangan berjalan mundur, contohnya `range(n_3002, 0, -1)`, yang digunakan untuk menghasilkan urutan angka dari nilai `n_3002` hingga 1.



### 6. Nested for

Nested `for` adalah penggunaan perulangan `for` yang ditempatkan di dalam perulangan `for` lainnya. Struktur ini memungkinkan program melakukan proses berulang secara bertingkat, terutama ketika data atau tampilan perlu disusun berdasarkan baris dan kolom.

Pada praktikum ini, nested `for` diterapkan untuk menghasilkan berbagai bentuk keluaran, seperti pola bintang, titik, deret angka, segitiga, dan jam pasir. Perulangan pertama biasanya berfungsi mengatur jumlah atau posisi baris, sedangkan perulangan di dalamnya digunakan untuk mengatur karakter atau nilai yang ditampilkan pada setiap baris.


### 7. Percabangan if, elif, dan else

Percabangan merupakan struktur program yang digunakan untuk menentukan perintah yang akan dijalankan sesuai dengan kondisi yang diberikan. Dalam Python, `if` digunakan untuk mengecek suatu kondisi, `elif` digunakan ketika terdapat kondisi lain yang perlu diperiksa, sedangkan `else` dijalankan apabila kondisi-kondisi sebelumnya tidak terpenuhi.

Pada praktikum ini, percabangan diterapkan untuk mengatur keluaran program sesuai dengan kondisi tertentu. Dalam pembuatan pola jam pasir, percabangan dapat digunakan untuk menentukan karakter atau angka yang ditampilkan berdasarkan posisi pada setiap baris dan kolom.



### 8. Operator Aritmatika 

Operator aritmatika digunakan untuk melakukan berbagai operasi perhitungan dalam program. Beberapa operator yang digunakan dalam praktikum ini antara lain:

* `+` untuk melakukan penjumlahan.
* `-` untuk melakukan pengurangan.
* `*` untuk melakukan perkalian.
* `%` untuk mendapatkan sisa hasil pembagian.

Operator tersebut digunakan dalam berbagai proses, seperti menghitung nilai, menentukan jumlah perulangan, mengatur posisi karakter, serta menentukan ukuran atau lebar pola yang akan ditampilkan.


### 9. Operator Perbandingan

Operator `in` termasuk operator membership yang digunakan untuk memeriksa apakah suatu nilai terdapat dalam sebuah kumpulan data, seperti list. Jika nilai yang dicari ditemukan, hasil pemeriksaan adalah `True`. Sebaliknya, jika nilai tersebut tidak terdapat di dalam kumpulan data, hasilnya adalah `False`.

Contoh penggunaannya:

`input_member_3002 in ["y", "ya"]`

Perintah tersebut digunakan untuk memeriksa masukan dari pengguna. Apabila pengguna memasukkan `y` atau `ya`, maka kondisi akan bernilai `True` dan dapat digunakan untuk menjalankan perintah tertentu.


### 10. Operator Logika or dan and

Operator or digunakan ketika salah satu dari beberapa kondisi sudah cukup untuk memenuhi suatu kondisi. Operator and digunakan ketika beberapa kondisi harus terpenuhi secara bersamaan.

Dalam pola jam pasir, operator or digunakan untuk memeriksa posisi tertentu seperti baris pertama atau baris terakhir, sedangkan and digunakan untuk menentukan posisi karakter di bagian dalam pola.

### 11. Parameter end pada print()

Fungsi `print()` pada Python secara otomatis membuat baris baru setelah menampilkan suatu nilai. Untuk mengatur agar keluaran tidak langsung berpindah baris, dapat digunakan parameter `end`.

Jika menggunakan `end=""`, hasil cetakan berikutnya akan tetap berada pada baris yang sama tanpa tambahan karakter. Sedangkan `end=" "` akan menjaga keluaran tetap dalam satu baris sekaligus memberikan satu spasi setelahnya.

Pengaturan tersebut sangat berguna dalam praktikum, terutama saat menyusun angka, karakter, dan spasi secara bertahap untuk menghasilkan pola sesuai dengan bentuk yang diinginkan.


### 12. Perulangan dan Akumulasi
Perulangan tidak hanya digunakan untuk menampilkan keluaran, tetapi juga dapat dimanfaatkan untuk melakukan perhitungan secara berulang. Salah satu penerapannya adalah menjumlahkan nilai pada setiap proses perulangan dengan menggunakan sebuah variabel sebagai penampung hasil.

Pada awal program, variabel penampung dapat diberikan nilai `0`. Kemudian, pada setiap perulangan, nilai tersebut ditambahkan dengan nilai dari variabel penghitung seperti `i`. Setelah seluruh proses perulangan selesai, nilai yang tersimpan pada variabel tersebut menjadi hasil penjumlahan dari semua proses yang telah dilakukan.


### 13. Pencetakan Pola

Pembuatan pola pada Python dapat dilakukan dengan memanfaatkan nested `for` untuk mengatur susunan karakter maupun angka. Bentuk pola ditentukan dengan mengatur banyaknya karakter yang dicetak, jumlah spasi, posisi angka, serta kapan program berpindah ke baris berikutnya.

Dalam praktikum ini, nested `for` digunakan untuk menghasilkan beberapa jenis pola, seperti susunan titik dan angka, bentuk persegi dengan karakter `*`, pola berdasarkan hasil perhitungan `i + j`, pola jam pasir, dan pola segitiga. Melalui latihan tersebut, dapat dipahami bagaimana perulangan digunakan untuk mengatur setiap bagian dari sebuah pola.



### 14. Pola Jam Pasir Kristal Palindromik

Pola jam pasir kristal palindromik dibuat dengan menggabungkan beberapa konsep pemrograman, seperti perulangan, percabangan, perhitungan, dan pencetakan karakter. Proses pembentukan pola dibagi menjadi tiga bagian, yaitu bagian atas, bagian tengah sebagai poros, dan bagian bawah.

Pada setiap baris, angka disusun dari nilai terbesar menuju angka 1, kemudian dilanjutkan kembali dari angka 1 menuju nilai terbesar sehingga membentuk susunan palindromik. Simbol `<*>` ditempatkan di tengah sebagai pusat pola. Sementara itu, karakter `#`, `=`, dan `|` digunakan untuk membuat bingkai yang mengelilingi pola.

Jumlah spasi pada bagian kiri dan kanan disesuaikan berdasarkan posisi baris. Pengaturan tersebut bertujuan agar setiap baris memiliki posisi yang tepat sehingga keseluruhan pola terlihat seimbang dan membentuk jam pasir yang simetris.

### 2.2.1 Program "jumlah_genap_2611533002"

````python

ulang_3002 = int(input("Masukkan nilai batas: "))

jumlah_3002 = 0
for i in range(1, ulang_3002 + 1):
    if i % 2 == 0:
        print(i, end=" ")
        jumlah_3002 = jumlah_3002 + i

        if i < ulang_3002:
            print(" + ", end="")
        else:
            print(" = ", jumlah_3002, end=" ")
print()
print("jumlah =", jumlah_3002)
````

**penjelasan :**

1. **Baris 1** berfungsi untuk meminta pengguna memasukkan nilai batas. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `ulang_3002`.

2. **Baris 2** berfungsi untuk memberikan nilai awal `0` pada variabel `jumlah_3002`. Variabel ini digunakan untuk menyimpan hasil penjumlahan bilangan genap.

3. **Baris 3** berfungsi untuk membuat perulangan menggunakan `for`. Variabel `i` akan berjalan dari angka `1` sampai nilai `ulang_3002`.

4. **Baris 4** berfungsi untuk memeriksa apakah nilai `i` merupakan bilangan genap. Operator `%` digunakan untuk mencari sisa pembagian `i` dengan `2`. Jika hasilnya `0`, maka nilai `i` adalah bilangan genap.

5. **Baris 5** berfungsi untuk mencetak nilai `i` yang merupakan bilangan genap. Parameter `end=" "` digunakan agar angka dicetak pada baris yang sama dan diberi spasi.

6. **Baris 6** berfungsi untuk menambahkan nilai `i` ke dalam variabel `jumlah_3002`. Hasil penjumlahan akan terus diperbarui setiap kali ditemukan bilangan genap.

7. **Baris 7** berfungsi untuk memeriksa apakah nilai `i` masih lebih kecil dari `ulang_3002`. Jika kondisi benar, program akan menjalankan perintah pada baris berikutnya.

8. **Baris 8** berfungsi untuk mencetak tanda `" + "`. Parameter `end=""` digunakan agar tanda tambah tetap berada pada baris yang sama.

9. **Baris 9** berfungsi sebagai kondisi alternatif menggunakan `else`. Kondisi ini dijalankan ketika nilai `i` sudah tidak lebih kecil dari `ulang_3002`.

10. **Baris 10** berfungsi untuk mencetak tanda `" = "` dan hasil penjumlahan yang tersimpan dalam variabel `jumlah_3002`.

11. **Baris 11** berfungsi untuk membuat baris baru setelah seluruh proses perulangan selesai.

12. **Baris 12** berfungsi untuk menampilkan hasil akhir dengan tulisan `"jumlah ="` dan nilai dari variabel `jumlah_3002`.


### 2.2.2 Program "nested_for1_2611533002"

````python
batas_3002 = int(input("Masukkan nilai batas: "))
for line_3002 in range(1, batas_3002 + 1):
    for j in range(1, (-1 * line_3002 + batas_3002) + 1):
        print(".", end="")
    print(line_3002) 
````

**Penjelasan:**

1. **Baris 1** berfungsi untuk meminta pengguna memasukkan nilai batas. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `batas_3002`.

3. **Baris 3** berfungsi untuk membuat perulangan di dalam perulangan luar menggunakan variabel `j`. Perulangan dimulai dari `1` sampai hasil perhitungan `(-1 * line_3002 + batas_3002) + 1`. Perulangan ini digunakan untuk menentukan jumlah titik yang dicetak pada setiap baris.

4. **Baris 4** berfungsi untuk mencetak karakter `"."` sebanyak jumlah perulangan `j`. Parameter `end=""` digunakan agar setiap titik dicetak pada baris yang sama tanpa berpindah baris.

5. **Baris 5** berfungsi untuk mencetak nilai `line_3002` setelah seluruh titik pada baris tersebut selesai dicetak. Perintah `print()` juga membuat program berpindah ke baris berikutnya setelah angka ditampilkan.

5. **Baris 5** berfungsi untuk membuat baris baru setelah seluruh hasil pada satu baris selesai dicetak. Perintah `print()` digunakan agar perulangan berikutnya dimulai pada baris baru.



### 2.2.5 Program "nested_for4_2611533002"

````python

tinggi_3002 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3002 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3002 = tinggi_3002
    c_3002 = a_3002
    lebar_3002 = (2 * tinggi_3002) - 2

    for i_3002 in range(1, tinggi_3002 + 1):
        b_3002 = c_3002 + 1

        for j_3002 in range(1, lebar_3002 + 1):

            #Baris atas dan bawah
            if i_3002 == 1 or i_3002 == tinggi_3002:
                 if j_3002 == 1 or j_3002 == lebar_3002:
                    print("#", end="")
                 else:
                    print("=", end="")
            #Baris isi
            else:
                if j_3002 == 1 or j_3002 == lebar_3002:
                    print("|", end="")
                else:
                    if j_3002 == c_3002:
                        print("<", end="")
                    elif j_3002 == b_3002:
                        print(">", end="")
                    elif j_3002 == (lebar_3002 - c_3002):
                        print("<", end="")
                    elif j_3002 == (lebar_3002 - c_3002 + 1):
                        print(">", end="")
                    elif j_3002 > b_3002 and j_3002 < (lebar_3002 - c_3002):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_3002 -= 2

        if a_3002 <= 0:
            c_3002 = (-a_3002) + 2
        else:
            c_3002 = a_3002
````

**Penjelasan:**
