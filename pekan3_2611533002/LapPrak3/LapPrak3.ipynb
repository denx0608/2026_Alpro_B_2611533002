## 2.1 Dasar Teori


## 2.1.1 Pengertian Operator

Operator adalah simbol atau kata yang digunakan untuk melakukan suatu proses pada data dalam program. Data yang digunakan bisa berupa angka, variabel, atau objek lainnya. Dalam Python, operator dapat digunakan untuk menghitung, membandingkan nilai, menggabungkan kondisi, dan mengolah data dalam bentuk biner.

Dalam praktikum ini, operator yang dipelajari meliputi operator aritmatika, assignment, perbandingan, logika, membership, identity, dan bitwise. Setiap operator memiliki fungsi yang berbeda dan dapat digunakan sesuai dengan kebutuhan dalam membuat program.


## 2.1.2 Operator Aritmatika

Operator aritmatika digunakan untuk melakukan perhitungan matematika dalam Python. Operator yang digunakan antara lain:

* `+` untuk penjumlahan
* `-` untuk pengurangan
* `*` untuk perkalian
* `/` untuk pembagian
* `//` untuk pembagian bilangan bulat
* `%` untuk mencari sisa pembagian
* `**` untuk perpangkatan

Pada program transaksi, operator aritmatika digunakan untuk menghitung diskon, total pembayaran, rata-rata harga barang, dan sisa pembagian jumlah barang.


# 2.1.3 Operator Assigment

Operator assignment digunakan untuk memasukkan atau mengubah nilai pada sebuah variabel. Operator dasarnya adalah `=`. Python juga menyediakan augmented assignment yang dapat digunakan untuk mengubah nilai variabel sekaligus melakukan operasi tertentu.

Contohnya:

```python
nilai_3002 = 25
nilai_3002 += 10
```

Setelah dijalankan, nilai `nilai_3002` menjadi **35**.

Beberapa operator assignment yang sering digunakan adalah `+=`, `-=`, `*=`, dan `/=`.


## 2.1.4 Operator Bitwise

Operator bitwise merupakan operator yang bekerja pada angka berdasarkan bentuk binernya. Operator ini biasanya digunakan ketika sebuah angka memiliki beberapa kondisi yang disimpan dalam bentuk bit.

Dalam praktikum ini digunakan beberapa operator bitwise, yaitu:

* `&` untuk mengecek bit tertentu
* `|` untuk menggabungkan bit
* `^` untuk membandingkan dua nilai bit
* `<<` untuk menggeser bit ke kiri
* `>>` untuk menggeser bit ke kanan

Pada studi kasus transaksi, setiap keadaan pelanggan memiliki nilai masing-masing. Pelanggan member menggunakan `0001`, belanja minimal Rp200.000 menggunakan `0010`, jumlah barang minimal 3 menggunakan `0100`, dan promo yang tersedia menggunakan `1000`.

Nilai-nilai tersebut kemudian dapat digabungkan menjadi satu kode status. Dengan cara ini, program dapat mengetahui kondisi pelanggan melalui satu nilai. Misalnya, operator `&` dapat digunakan untuk mengecek apakah pelanggan merupakan member, sedangkan operator `|` dapat menggabungkan beberapa kondisi sekaligus.




## 2.1.5 Operator Logika

Operator logika digunakan untuk menggabungkan beberapa kondisi dalam sebuah program. Python memiliki tiga operator logika yang sering digunakan, yaitu `and`, `or`, dan `not`.

Operator `and` akan menghasilkan `True` jika semua kondisi bernilai `True`. Operator `or` menghasilkan `True` jika salah satu kondisi bernilai `True`. Sedangkan `not` digunakan untuk membalik nilai Boolean.

Contohnya:

```python
member_3002 = True
belanja_3002 = True

hasil_3002 = member_3002 and belanja_3002
```

Hasil dari operasi tersebut adalah **True** karena kedua kondisi memiliki nilai `True`.




## 2.1.6 Operator Perbandingan

Operator perbandingan digunakan untuk melihat hubungan antara dua nilai. Hasil yang diberikan berupa `True` atau `False`, tergantung pada kondisi yang dibandingkan.

Beberapa operator perbandingan yang digunakan dalam Python yaitu:

* `==` untuk mengecek apakah nilainya sama
* `!=` untuk mengecek apakah nilainya berbeda
* `>` untuk mengecek nilai lebih besar
* `<` untuk mengecek nilai lebih kecil
* `>=` untuk mengecek nilai lebih besar atau sama dengan
* `<=` untuk mengecek nilai lebih kecil atau sama dengan

Pada program transaksi, operator ini digunakan untuk mengecek apakah total belanja sudah mencapai Rp200.000 dan apakah jumlah barang sudah mencapai minimal 3 barang.



## 2.1.7 Operator Keanggotaan

Membership merupakan operator pada Python yang digunakan untuk mencari tahu apakah sebuah data termasuk dalam suatu kumpulan nilai. Pengecekan ini dapat dilakukan pada beberapa jenis data, seperti list, tuple, maupun string.

Python menyediakan dua operator untuk melakukan pengecekan tersebut, yaitu `in` dan `not in`. Operator `in` akan menghasilkan `True` jika data yang dicari ditemukan. Sebaliknya, `not in` menghasilkan `True` jika data tersebut tidak ditemukan.

Sebagai contoh:

```python
kode_3002 = "SKIRK06"
daftar_3002 = ["FURINA09", "YELAN08", "SKIRK06"]

cek_3002 = kode_3002 in daftar_3002
```

Pada contoh tersebut, hasil `cek_3002` adalah `True` karena `"SKIRK06"` termasuk salah satu data yang ada di dalam list `daftar_3002`.



## 2.1.8 Operator Identitas

Operator identity pada Python berfungsi untuk mengetahui apakah dua variabel menunjuk ke objek yang sama di dalam memori. Jadi, yang diperiksa bukan hanya isi datanya, tetapi apakah keduanya benar-benar menggunakan objek yang sama.

Ada dua operator identity, yaitu:

* `is` digunakan untuk mengecek apakah dua variabel merupakan objek yang sama.
* `is not` digunakan untuk mengecek apakah dua variabel merupakan objek yang berbeda.

Operator `is` berbeda dengan `==`. Operator `is` melihat identitas objek, sedangkan `==` hanya membandingkan nilai yang terdapat di dalam objek.

Contohnya, dua list bisa saja memiliki isi yang sama, tetapi dibuat sebagai objek yang berbeda. Dalam kondisi tersebut, `==` dapat menghasilkan `True`, sedangkan `is` menghasilkan `False`.

````python
# Buat prograam file dengan nama aritmatika_2611533002.py
# Buat program untuk operator aritmatika dalam Python
# Buat variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3002 = int(input("Masukkan angka angka-1: "))
angka2_3002 = int(input("Masukkan angka angka-2: "))

# Penjumlahan
hasil = angka1_3002 + angka2_3002
print("\nOperator penjumlahan")
print("Hasil =", hasil)    

# Pengurangan
hasil = angka1_3002 - angka2_3002
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian
hasil = angka1_3002 * angka2_3002
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian
if angka2_3002 != 0:
    hasil = angka1_3002 / angka2_3002
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_3002 // angka2_3002
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_3002 % angka2_3002
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh vernilai 0.")

# Pangkat
hasil = angka1_3002 ** angka2_3002
print("\nOperator Pangkat")
print("Hasil =", hasil)
````

### 2.2.2 Program "assigment_2611533002"

````python
# bualah file dengan nama assigment_2611533002.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_3002 = int(input("Input angka angka-1: "))
angka2_3002 = int(input("Input angka angka-2: "))

print("nNilai awal angka1 =", angka1_3002)
print("Nilai awal angka2 =", angka2_3002)

# Assigment biasa
hasil_3002 = angka1_3002
hasil_3002 += angka2_3002
print("\nAssignment penambahan (+=)")
print("hasil =", hasil_3002)

# Assignment penambahan
hasil = angka1_3002
hasil += angka2_3002
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil_3002 = angka1_3002
hasil_3002 -= angka2_3002
print("\nAssignment pengurangan (-=)")
print("hasil =", hasil_3002)

# Assignment perkalian
hasil_3002 = angka1_3002
hasil_3002 *= angka2_3002
print("\nAssignment perkalian (*=)")
print("hasil =", hasil_3002)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3002 != 0:
    hasil_3002 = angka1_3002
    hasil_3002 /= angka2_3002
    print("\nAssignment pembagian (/=)")
    print("hasil =", hasil_3002)

    #Operator tambahan
    hasil_3002 = angka1_3002
    hasil_3002 //= angka2_3002
    print("\nAssignment pembagian bulat (//=)")
    print("hasil =", hasil_3002)
    hasil_3002 = angka1_3002
    hasil_3002 %= angka2_3002
    print("\nAssignment sisa bagi (%=)")
    print("hasil =", hasil_3002)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_3002 = angka1_3002
hasil_3002 **= angka2_3002
print("\nAssignment perpangkatan (**=)")
print("hasil =", hasil_3002)
````

### 2.2.3 Program "bitwise_2611533002"

````python
# Buatlah file dengan nama bitwise_2611533002.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator bitwise dalam Python

print("\n==================================")
print("3. OPERATOR BITWISE")
print("==================================")

angka1_3002 = int(input("Masukkan angka bitwise-1: "))
angka2_3002 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner:")
print("Angka 1:", angka1_3002, "-| biner", bin(angka1_3002))
print("Angka 2:", angka2_3002, "-| biner", bin(angka2_3002))

# Bitwise AND
hasil_3002 = angka1_3002 & angka2_3002
print("\nBitwise AND (&)")
print(angka1_3002, "&", angka2_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))

# Bitwise OR
hasil_3002 = angka1_3002 | angka2_3002
print("\nBitwise OR (|)")
print(angka1_3002, "|", angka2_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))

# Bitwise XOR
hasil_3002 = angka1_3002 ^ angka2_3002
print("\nBitwise XOR (^)")
print(angka1_3002, "^", angka2_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))

# Bitwise NOT
hasil_3002 = ~angka1_3002
print("\nBitwise NOT (~)")
print("~", angka1_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))

# Bitwise geser kiri
jumlah_geser_3002 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3002 = angka1_3002 << jumlah_geser_3002
print("\nBitwise Geser Kiri (<<)")
print(angka1_3002, "<<", jumlah_geser_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))

# Bitwise geser kanan
hasil_3002 = angka1_3002 >> jumlah_geser_3002
print("\nBitwise Geser Kanan (>>)")
print(angka1_3002, ">>", jumlah_geser_3002, "=", hasil_3002)
print("Biner hasil =", bin(hasil_3002))
print("Biner hasil (8 bit) =", format(hasil_3002, '08b'))
````

5. **Line 24–28** digunakan untuk operasi **Bitwise OR (`|`)**. Operator OR menghasilkan bit `1` jika salah satu atau kedua bit bernilai `1`. Hasilnya ditampilkan dalam bentuk desimal dan biner.

Setelah melakukan praktikum Algoritma dan Pemrograman Pekan 3, dapat dipahami bahwa operator sangat dibutuhkan dalam membuat program Python. Setiap operator mempunyai kegunaan yang berbeda. Operator aritmatika digunakan untuk melakukan perhitungan, assignment untuk mengatur atau mengubah nilai, perbandingan untuk melihat hubungan antara dua nilai, dan operator logika untuk menggabungkan beberapa kondisi.

Selain itu, terdapat operator `membership` yang digunakan untuk mengecek apakah suatu data ada di dalam sebuah kumpulan, serta operator `identity` untuk mengetahui apakah dua variabel mengarah pada objek yang sama. Operator `bitwise` juga dipelajari untuk mengolah nilai dalam bentuk biner dan menggabungkan beberapa kondisi menjadi sebuah kode tertentu.

Pada studi kasus simulasi transaksi dan validasi akses toko, semua operator tersebut dicoba dalam satu program. Program dapat menerima data pelanggan, menghitung transaksi, mengecek diskon dan kode promo, serta menentukan hak akses berdasarkan kondisi yang dimiliki pelanggan.

Dari praktikum ini, pemahaman mengenai berbagai operator Python menjadi lebih jelas karena tidak hanya mempelajari teori, tetapi juga langsung menerapkannya dalam sebuah program. Praktikum ini juga membantu melatih cara berpikir secara lebih teratur dalam membuat program untuk menyelesaikan suatu permasalahan.
