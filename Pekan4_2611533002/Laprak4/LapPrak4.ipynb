## 2.1 Dasar Teori

## 2.1.1 Struktur Percabangan ##

Struktur percabangan adalah bagian penting dalam pemrograman yang berfungsi untuk menentukan tindakan program berdasarkan suatu kondisi. Melalui percabangan, program dapat memilih perintah yang akan dijalankan sesuai dengan situasi atau data yang diperoleh. Pada Python, terdapat beberapa bentuk percabangan yang sering digunakan, seperti `if`, `if-else`, `if-elif-else`, multi-if, dan `match-case`.


## 2.1.2 pernyataan if

Pernyataan `if` berfungsi untuk menjalankan sebuah perintah ketika kondisi yang diperiksa menghasilkan nilai benar (`True`). Jika kondisi tersebut tidak terpenuhi atau bernilai salah (`False`), maka perintah di dalam `if` akan dilewati dan program meneruskan proses ke bagian selanjutnya.

Bentuk dasar penggunaan `if` adalah:

`if kondisi:`

`    perintah`

Pada program yang menggunakan nilai IPK, `if` dapat digunakan untuk mengecek apakah IPK seseorang sudah mencapai nilai minimum yang ditentukan. Percabangan ini biasanya digunakan jika program hanya perlu melakukan pemeriksaan terhadap satu kondisi.


### 2.1.3 Pernyataan if-else

Pernyataan `if-else` digunakan untuk menangani kondisi yang memiliki dua pilihan hasil. Ketika kondisi pada `if` terpenuhi atau bernilai `True`, program akan menjalankan perintah yang terdapat pada bagian `if`. Namun, apabila kondisi tersebut bernilai `False`, program akan menjalankan perintah pada bagian `else`.

Bentuk umum:

`if kondisi:`

`    perintah jika benar`

`else:`

`    perintah jika salah`

Dalam contoh program yang menggunakan nilai IPK, `if-else` dapat diterapkan untuk menentukan status pengguna, seperti apakah pengguna dinyatakan lulus atau tidak berdasarkan nilai IPK yang diperoleh.


## 2.1.4 Pernyataan if-elif-else

Pernyataan `if-elif-else` digunakan untuk program yang mempunyai beberapa kondisi yang perlu diperiksa. Program akan mengecek kondisi mulai dari `if`, kemudian dilanjutkan ke `elif` apabila kondisi sebelumnya tidak terpenuhi. Jika tidak ada kondisi yang sesuai, maka bagian `else` akan dijalankan. Setelah menemukan kondisi yang benar, program langsung menjalankan perintah pada bagian tersebut dan tidak melanjutkan pengecekan kondisi lainnya.

Bentuk umum:

`if kondisi_1:`

`    perintah_1`

`elif kondisi_2:`

`    perintah_2`

`else:`

`    perintah_3`

Pada program pemeriksaan umur dan kepemilikan SIM C, struktur ini digunakan untuk menentukan apakah seseorang diperbolehkan mengendarai motor berdasarkan persyaratan yang telah ditentukan. Selain itu, `if-elif-else` juga dapat digunakan untuk mengubah nilai angka menjadi nilai huruf sesuai dengan rentang nilai yang diperoleh.


## 2.1.5 Multi-if

Multi-if merupakan penggunaan beberapa pernyataan `if` dalam satu program dengan kondisi yang berdiri sendiri. Pada struktur ini, setiap kondisi akan diperiksa tanpa bergantung pada kondisi lainnya. Oleh karena itu, beberapa kondisi dapat terpenuhi sekaligus dan perintah dari masing-masing kondisi yang benar akan tetap dijalankan.

Penggunaan multi-if sesuai untuk situasi ketika satu pengguna atau data dapat memenuhi beberapa persyaratan dalam waktu yang sama. Salah satu contohnya adalah program pemberian diskon, di mana pelanggan dapat memperoleh diskon berdasarkan jumlah belanja, status member, dan penggunaan kode promo. Karena setiap kondisi diperiksa secara terpisah menggunakan `if`, diskon yang memenuhi syarat dapat dijumlahkan menjadi satu total diskon.


## 2.1.6 Operator Perbandingan dan Operator logika3

Operator perbandingan berfungsi untuk mengetahui hubungan antara dua buah nilai. Hasil dari perbandingan tersebut berupa `True` jika kondisi terpenuhi atau `False` jika kondisi tidak terpenuhi. Beberapa operator perbandingan yang digunakan dalam program yaitu:

* `>` untuk menentukan apakah nilai pertama lebih besar daripada nilai kedua.
* `>=` untuk mengecek apakah nilai pertama lebih besar atau sama dengan nilai kedua.
* `<` untuk mengecek apakah nilai pertama lebih kecil daripada nilai kedua.
* `!=` untuk mengetahui apakah dua nilai memiliki nilai yang berbeda.
* `==` untuk mengecek apakah dua nilai memiliki nilai yang sama.

Selain operator perbandingan, terdapat operator logika yang dapat digunakan untuk menggabungkan dua atau lebih kondisi. Salah satunya adalah `and`, yang akan menghasilkan `True` apabila semua kondisi yang diberikan terpenuhi.

Contoh pada pengecekan umur dan SIM C:

`umur_3003 >= 17 and sim_3002 == 'y'`

Kondisi tersebut akan menghasilkan `True` apabila umur pengguna sudah mencapai 17 tahun atau lebih serta pengguna memiliki SIM C.


## 2.1.7 Input Konversi Tipe Data

Fungsi `input()` digunakan untuk mengambil data yang diberikan oleh pengguna melalui keyboard. Data yang diperoleh dari `input()` secara otomatis memiliki tipe `string`, sehingga perlu dilakukan konversi jika ingin digunakan sebagai tipe data lain.

Konversi dapat dilakukan menggunakan fungsi `int()` untuk mengubah data menjadi bilangan bulat dan `float()` untuk mengubahnya menjadi bilangan desimal.

Contohnya:

`umur_3002 = int(input("Input umur anda: "))`

Pada contoh tersebut, nilai umur yang dimasukkan pengguna dikonversi menjadi tipe integer. Dengan begitu, data tersebut dapat digunakan untuk melakukan perhitungan maupun perbandingan dalam program.


## 2.1.8 Penggunaan strip(), lower(), dan [0]

Metode `.strip()` berfungsi untuk menghilangkan spasi yang tidak diperlukan pada bagian awal dan akhir sebuah teks. Sementara itu, `.lower()` digunakan untuk mengubah semua huruf pada teks menjadi huruf kecil. Kedua metode ini berguna untuk membuat data input dari pengguna lebih mudah disesuaikan dengan kondisi yang ada di dalam program.

Contohnya:

`input_member_3002 = input("Apakah Anda member? (y/t): ").strip().lower()`

Selain itu, indeks `[0]` digunakan untuk mengambil karakter pertama dari sebuah string. Dalam program pengecekan SIM C, penggunaan `[0]` dapat digunakan untuk mengambil huruf pertama dari input pengguna, seperti `y` atau `t`.


## 2.1.9 Operator in

Operator `in` digunakan untuk mengecek apakah suatu nilai terdapat di dalam kumpulan data, misalnya list. Hasil yang diberikan berupa `True` jika nilai yang dicari ditemukan, sedangkan jika tidak ada maka hasilnya `False`.

Contohnya:

`input_member_3002 in ["y", "ya"]`

Kode tersebut digunakan untuk mengecek input pengguna. Jika pengguna memasukkan `y` atau `ya`, maka kondisi akan menghasilkan `True`.


## 2.1.10 match-case

`match-case` adalah salah satu bentuk percabangan pada Python yang digunakan untuk mencocokkan nilai dengan beberapa pilihan yang telah ditentukan. Struktur ini dapat mempermudah penulisan program ketika terdapat banyak pilihan atau kemungkinan yang perlu diproses.

Bentuk umum:

`match variabel:`

`    case nilai_1:`

`        perintah`

`    case nilai_2:`

`        perintah`

`    case _:`

`        perintah_default`

Dalam contoh program untuk mengubah angka menjadi nama bulan, `match-case` digunakan untuk mencocokkan angka dari 1 hingga 12 dengan bulan yang sesuai. Sementara itu, `case _` berfungsi sebagai pilihan bawaan yang dijalankan apabila nilai yang dimasukkan tidak sesuai dengan pilihan yang tersedia.


## 2.1.11 Operasi Perhitungan Diskon

Pada program untuk menghitung diskon, operator aritmatika digunakan untuk menentukan besar potongan harga dan jumlah uang yang harus dibayar. Nilai persentase diskon terlebih dahulu dikonversi ke bentuk desimal dengan cara dibagi 100.

Rumus yang digunakan yaitu:

**Nominal Diskon = Total Belanja × (Total Diskon / 100)**

**Total Bayar = Total Belanja - Nominal Diskon**

Jika program menggunakan konsep multi-if, beberapa jenis diskon yang berhasil memenuhi syarat dapat dikumpulkan terlebih dahulu. Setelah total persentase diskon diperoleh, nilai tersebut digunakan untuk menghitung jumlah potongan dan total pembayaran.


## 2.1.12 Konversi Nilai Angka Menjadi Nilai Huruf

Struktur `if-elif-else` dapat dimanfaatkan untuk menentukan kategori berdasarkan nilai yang diperoleh. Dalam program ini, nilai angka yang dimasukkan akan diperiksa dan kemudian diubah menjadi nilai huruf sesuai dengan rentang yang telah ditentukan.

Jika nilai yang diperoleh 81 ke atas, maka termasuk kategori A. Nilai 70–80 masuk kategori B, 60–69 masuk kategori C, 50–59 masuk kategori D, sedangkan nilai kurang dari 50 termasuk kategori E.

Kondisi tersebut diperiksa secara bertahap sehingga program hanya menghasilkan satu kategori yang sesuai dengan nilai input pengguna.


## 2.1.13 Kesimpulan Dasar Teori

Dari pembahasan yang telah dijelaskan, dapat diketahui bahwa struktur percabangan memiliki peran penting dalam mengatur jalannya sebuah program. Percabangan memungkinkan program menentukan tindakan yang sesuai berdasarkan kondisi yang diberikan. `if` digunakan untuk memeriksa satu kondisi, `if-else` menangani dua pilihan, sedangkan `if-elif-else` digunakan ketika terdapat beberapa kondisi yang harus diperiksa secara berurutan. Sementara itu, multi-if memungkinkan setiap kondisi diperiksa secara terpisah, dan `match-case` digunakan untuk memilih tindakan berdasarkan nilai yang cocok dengan pilihan tertentu.

Selain struktur percabangan, penggunaan operator perbandingan dan operator logika juga membantu dalam menentukan hasil dari suatu kondisi. Fungsi `input()` digunakan untuk menerima data dari pengguna, sedangkan konversi tipe data diperlukan agar data tersebut dapat diproses sesuai kebutuhan program.

Penerapan konsep-konsep tersebut dapat ditemukan dalam berbagai contoh permasalahan, seperti mengecek kelayakan seseorang, menentukan kategori nilai, menghitung potongan harga, hingga mengubah angka menjadi nama bulan. Dengan memahami konsep ini, pembuatan program dengan alur keputusan dapat dilakukan dengan lebih teratur dan mudah dipahami.

## 2.2.3 Program "if_elif_else1_2611533002"

6. Baris 13 berfungsi menampilkan pesan bahwa pengguna sudah dewasa, tetapi belum diperbolehkan membawa motor karena tidak memiliki SIM C.

## 2.2.4 Program "match_case_2611533002"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_3002 = int(input("Masukan angka bulan (1- 12): "))

match bulan_3002:
    case 1:
        print("januari")
    case 2:
        print("Februari")
    case 3:
        print("maret")
    case 4:
        print("april")
    case 5:
        print("mei")
    case 6:
        print("juni")
    case 7:
        print("juli")
    case 8:
        print("agustus")
    case 9:
        print("september")
    case 10:
        print("oktober")
    case 11:
        print("november")
    case 12:
        print("desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan:**

1. Baris 7 berfungsi untuk meminta pengguna memasukkan angka bulan dari 1 sampai 12. Fungsi int() digunakan untuk mengubah input menjadi bilangan bulat, kemudian disimpan dalam variabel bulan_3002.

2. Baris 9 menggunakan match untuk mencocokkan nilai yang terdapat pada variabel bulan_3002 dengan pilihan yang tersedia.

## 2.2.6 Program "multi_if2_2611533002"

```python
# Buat file dengan nama multi_if2_2611533002.py
# Buat program untuk kondisional if
# Nama variable ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Porgoram menghitung diskon belanja

# Input dari user
total_belanja_3002 = float(input("Masukan total belanja (Rp): "))

# input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3002 = input("Apakah anda member? (y/t) :").strip().lower()
is_member_3002 = input_member_3002 in ["y", "ya"]

# input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3002 = input("Apakah kode promo valid? (y/t): ").strip().lower()
Kode_promo_valid_3002 = input_promo_3002 in ["y", "ya"]

total_diskon_persen_3002 = 0

# Multi-if terpisah setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3002 > 1000000:
    total_diskon_persen_3002 += 10 # Diskon belanja besar

if is_member_3002:
    total_diskon_persen_3002 += 5 # Diskon member

if Kode_promo_valid_3002:
    total_diskon_persen_3002 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3002 = total_belanja_3002 * (total_diskon_persen_3002 / 100)
total_bayar_3002 = total_belanja_3002 - nominal_diskon_3002

#output hasil
print("n--- rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_3002}% (Rp {nominal_diskon_3002:,.0f})")
print(f"Total bayar  : Rp {total_bayar_3002:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3002}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member dan kode promo valid
```

**Penjelasan:**

1. Baris 7 berfungsi meminta pengguna memasukkan total belanja. `float()` digunakan agar nilai bisa berupa angka desimal, lalu disimpan ke variabel `total_belanja_3002`.

2. Baris 10 meminta pengguna memasukkan status member dengan pilihan `y/t`. `.strip()` menghapus spasi tambahan dan `.lower()` mengubah input menjadi huruf kecil.

3. Baris 11 mengecek apakah input member adalah `y` atau `ya`. Jika iya, `is_member_3002` bernilai `True`.

4. Baris 14 meminta pengguna memasukkan apakah kode promo valid. `.strip().lower()` digunakan agar input lebih mudah diproses.

5. Baris 15 mengecek apakah kode promo yang dimasukkan adalah `y` atau `ya`. Jika iya, `Kode_promo_valid_3002` bernilai `True`.

6. Baris 17 membuat variabel `total_diskon_persen_3002` dengan nilai awal `0%`.

7. Baris 22 mengecek apakah total belanja lebih dari Rp1.000.000. Jika iya, diskon ditambah `10%`.

8. Baris 25 mengecek apakah pembeli merupakan member. Jika iya, diskon ditambah `5%`.

9. Baris 28 mengecek apakah kode promo valid. Jika iya, diskon ditambah `15%`.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# Input data pengunjung
nama_3002 = input("Masukkan Nama Pengunjung        : ")
umur_3002 = int(input("Input umur anda                 : "))
sim_3002 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]
jumlah_tiket_3002 = int(input("Masukkan jumlah tiket           : "))

# If tunggal untuk validasi jumlah tiket
if jumlah_tiket_3002 <= 0:
    print("Peringatan: Kuota tiket tidak valid.")

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_3002 = int(input("Masukkan nomor paket (1-5)      : "))

# Match-case untuk memilih wahana
match paket_3002:
    case 1:
        nama_wahana_3002 = "Wahana Safari Rimba"
        harga_satuan_3002 = 50000

    case 2:
        nama_wahana_3002 = "Wahana Arung Jeram"
        harga_satuan_3002 = 75000

    case 3:
        nama_wahana_3002 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3002 = 120000

    case 4:
        nama_wahana_3002 = "Wahana Roller Coaster Kilat"
        harga_satuan_3002 = 100000

    case 5:
        nama_wahana_3002 = "Wahana All-Access VIP"
        harga_satuan_3002 = 220000

    case _:
        print("Paket wahana tidak valid!")
        exit()

# Input member dan promo
is_member_3002 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_3002 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# Validasi izin wahana
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3002 == 3:
    if umur_3002 >= 17 and sim_3002 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_3002 >= 17 and sim_3002 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_3002 < 17 and sim_3002 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_3002 >= 10:
        print("Status Akses: Anda memenuhi syarat umur untuk wahana.")
    else:
        print("Status Akses: Anda belum cukup umur untuk wahana ini.")

# Perhitungan subtotal
subtotal_3002 = harga_satuan_3002 * jumlah_tiket_3002

# Multi-if terpisah untuk diskon akumulatif
total_diskon_persen_3002 = 0

if subtotal_3002 >= 200000:
    total_diskon_persen_3002 += 10

if is_member_3002 in ['y', 'ya']:
    total_diskon_persen_3002 += 5

if kode_promo_valid_3002 in ['y', 'ya']:
    total_diskon_persen_3002 += 15

if jumlah_tiket_3002 >= 5:
    total_diskon_persen_3002 += 5

# Perhitungan pembayaran
nominal_diskon_3002 = subtotal_3002 * (total_diskon_persen_3002 / 100)
total_bayar_3002 = subtotal_3002 - nominal_diskon_3002

# Audit transaksi
if total_bayar_3002 > 300000:
    catatan_layanan_3002 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_3002 = "Terima kasih telah berkunjung."

# Menampilkan rincian
print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Nama Pengunjung  : {nama_3002}")
print(f"Wahana           : {nama_wahana_3002}")
print(f"Subtotal Belanja : Rp {subtotal_3002:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3002}% (Rp {nominal_diskon_3002:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3002:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_3002}")

print("Program Selesai")
```
