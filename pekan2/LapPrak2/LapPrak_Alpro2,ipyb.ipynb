## 2.1 Dasar Teori
1. Mengenal Bahasa Pemrograman Python

Python adalah bahasa pemrograman tingkat tinggi yang bersifat interpreted, object-oriented, dan
dynamically typed. Bahasa ini dikembangkan oleh Guido van Rossum dan pertama kali diperkenalkan pada
tahun 1991. Python memiliki bentuk penulisan yang relatif sederhana sehingga mudah dibaca dan dipelajari.
Penggunaannya cukup luas, mulai dari pembuatan aplikasi, otomatisasi pekerjaan, pengolahan data, hingga
kecerdasan buatan.

2. Variabel pada Python

Variabel dapat dianggap sebagai tempat untuk menaruh sebuah nilai agar nilai tersebut dapat digunakan di
dalam program. Pada Python, variabel bisa langsung dibuat dengan menggunakan tanda `=` tanpa
menuliskan tipe data terlebih dahulu. Ada beberapa aturan dalam membuat nama variabel, misalnya huruf
besar dan kecil dianggap berbeda, nama variabel tidak boleh diawali angka, dan kata yang sudah menjadi
keyword Python tidak boleh digunakan sebagai nama variabel.
3. Tipe Data Dasar

Tipe data menunjukkan bentuk atau jenis nilai yang terdapat pada suatu variabel. Beberapa tipe data dasar
yang sering digunakan dalam Python antara lain:
• **Integer (`int`)**: untuk bilangan bulat, contohnya `-5`, `0`, dan `100`.
• **Float (`float`)**: untuk bilangan yang mempunyai angka di belakang koma, contohnya `3.14` dan `-0.01`.
• **String (`str`)**: untuk menyimpan teks atau kumpulan karakter yang ditulis di dalam tanda petik,
contohnya `"Praktikum Python"`.
• **Boolean (`bool`)**: untuk menyimpan nilai logika, yaitu `True` atau `False`.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611533002.py"


```python
# Buat file dengan nama Boolean_2611533002.py
# Nama variable ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasikan variable dengan type Boolean
is_lulus_3002 = True
is_cumlaude_3002 = True

# Menggunakan Boolean
nilai_3002 = 80
batas_lulus_3002 = 75

#Menentukan nilai Boolean dari kondisi
status_kelulusan_3002 = nilai_3002 >= batas_lulus_3002 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai", nilai_3002)
print("Apakah Lulus?", status_kelulusan_3002)
if is_lulus_3002 and is_cumlaude_3002:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :**
1. Dua variabel pertama berisi nilai Boolean untuk menunjukkan status lulus dan status cumlaude.
2. `nilai_3002` menyimpan nilai mahasiswa, sedangkan `batas_lulus_3002` menyimpan nilai minimum untuk
dinyatakan lulus.
3. Perbandingan kedua nilai tersebut menghasilkan `True` atau `False` pada variabel status kelulusan.
4. Hasil akhirnya ditampilkan ke layar menggunakan beberapa perintah `print()`.

### 2.2.2 Program Driver "Input_2611533002.py"

```python
# Buat file dengan nama Input_2611533002.py
# Buat Program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasaukan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukin akan dikonversi menjadi tipe data integer
panjang_3002 = input("Masukan panjang: ")
lebar_3002 = input("Masukan lebar: ")
luas_3002 = int(panjang_3002) * int(lebar_3002)
print("luas = ", luas_3002)
```

**Penjelasan:**

 1. Program meminta pengguna mengisi panjang dan lebar persegi panjang melalui `input()`.
2. Kedua input diubah menjadi bilangan integer, lalu dikalikan untuk mendapatkan luas.
3. Hasil luas yang sudah dihitung ditampilkan pada layar.

### 2.2.3 Program "Konstanta_2611533002.py"

```python
# Buat file dengan nama Konstanta_2611533002.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variable ditambah 4 digit nim terakhir contoh: phi_1234

from typing import Final
PI : Final = 3.14
print("pi: %f" % (PI))
jari_3002 = float(input("Masukan jari-jari lingkaran: "))
luas_3002 = PI * jari_3002 * jari_3002
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f"% (jari_3002, luas_3002))
```

**Penjelasan:**

1. Program mengimpor `Final` dan menetapkan `PI` dengan nilai 3.14 sebagai konstanta.
2. Nilai PI ditampilkan, kemudian pengguna diminta memasukkan jari-jari lingkaran.
3. Nilai jari-jari digunakan bersama PI untuk menghitung luas lingkaran.
4. Hasil perhitungan ditampilkan dengan dua angka di belakang koma.

### 2.2.4 Program "Numerik_261153302.py"

```python
# Buat dile dengan nama Numerik_22611533002.py
#create a variabel with integer float value
a_3002=100
print("The type of variable having value", a_3002, " is ", type(a_3002))

#create a variable with float value.0
b_3002=10.2345
print("The type of variable having valeu", b_3002, " is ", type(b_3002))

#create a variable with complex value.

c_3002=100+3j
print("The type of variable having value", c_3002, " is ", type(c_3002))
```

**Penjelasan:**

1. Variabel `a_3002` diisi dengan bilangan bulat 100, lalu tipe datanya diperiksa menggunakan `type()`.
2. Variabel `b_3002` berisi nilai desimal 10.2345 dan program menampilkan tipe datanya.
3. Variabel `c_3002` berisi bilangan kompleks `100+3j`, kemudian tipe datanya juga diperiksa.
